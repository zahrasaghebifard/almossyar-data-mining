```python
import pandas as pd
import re
from pathlib import Path

from difflib import SequenceMatcher
```

```python
df = pd.read_csv("../data/processed/messages_parsed.csv")

print(df.shape)
df.head()
```

```python
def normalize_persian(text):
    if not isinstance(text, str):
        return ""
    text = text.replace("ي", "ی").replace("ك", "ک")
    text = text.replace("‌", " ")  # نیم‌فاصله
    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()
```

```python
df["professor_name_clean"] = (
    df["professor_name"]
    .fillna("")
    .apply(normalize_persian)
    .str.lower()
)

df[["professor_name", "professor_name_clean"]].head(10)
```

```python
def similarity(a, b):
    return SequenceMatcher(None, a, b).ratio()
```

```python
df["professor_id"] = df["professor_name_clean"].factorize()[0]
```

```python
df["course_name_clean"] = (
    df["course"]
    .fillna("")
    .apply(normalize_persian)
    .str.lower()
)

# مثال ساده برای اعداد
df["course_name_clean"] = (
    df["course_name_clean"]
    .str.replace("۲", "2")
    .str.replace("۱", "1")
)

df[["course", "course_name_clean"]].head(10)
```

```python
def map_grading(x):
    if not isinstance(x, str):
        return "نامشخص"
    if "سخت" in x:
        return "سختگیر"
    if "آسان" in x or "راحت" in x:
        return "آسان"
    if "منصف" in x:
        return "منصفانه"
    return "نامشخص"


df["grading"] = df["grading"].apply(map_grading)

df["grading"].value_counts()
```

```python
def map_attendance(x):
    if not isinstance(x, str):
        return "نامشخص"
    if "سخت" in x:
        return "سختگیر"
    if "آزاد" in x:
        return "آزاد"
    if "متوسط" in x:
        return "متوسط"
    return "نامشخص"


df["attendance"] = df["attendance"].apply(map_attendance)

df["attendance"].value_counts()
```

```python
df["comment_clean"] = (
    df["raw_text"]
    .fillna("")
    .apply(normalize_persian)
)

df[["raw_text", "comment_clean"]].head(5)
```

```python
df.isna().mean().sort_values(ascending=False)
```

```python
# ---------- clean text fields ----------
df["professor_name_clean"] = (
    df["professor_name"]
    .fillna("")
    .apply(normalize_persian)
)

df["course_name_clean"] = (
    df["course"]
    .fillna("")
    .apply(normalize_persian)
)

df["comment_clean"] = (
    df["raw_text"]
    .fillna("")
    .apply(normalize_persian)
)

# ---------- ids & meta ----------
df["professor_id"] = df["professor_name_clean"].astype("category").cat.codes

# اگر از JSON تاریخ نگرفتیم
df["id"] = range(1, len(df) + 1)
df["date"] = pd.NaT
df["date_unixtime"] = None

# ---------- ratings (from evaluation columns) ----------
RATING_MAP = [
    "eval_پیوستگی و یکپارچگی تدریس",
    "eval_دانش عمومی",
    "eval_توانایی انتقال مطالب و نکات مهم درس",
    "eval_نحوه مدیریت کلاس(نظم و زمان)",
    "eval_پاسخگویی(حضوری و غیرحضوری)",
    "eval_آداب و رفتار اجتماعی با دانشجویان"
]

for i, col in enumerate(RATING_MAP, start=1):
    df[f"rating_{i}"] = df[col] if col in df.columns else None

# ---------- rename status fields ----------
df["grading_status"] = df["grading"]
df["attendance_status"] = df["attendance"]

# ---------- final dataframe ----------
final_columns = [
    "id", "date", "date_unixtime",
    "professor_id", "professor_name_clean",
    "department",
    "course_name_clean",
    "rating_1", "rating_2", "rating_3",
    "rating_4", "rating_5", "rating_6",
    "grading_status",
    "attendance_status",
    "comment_clean"
]

df_clean = df[final_columns]

df_clean.head()
```

```python
Path("../data/processed").mkdir(exist_ok=True)

df_clean.to_csv(
    "../data/processed/messages_clean.csv",
    index=False,
    encoding="utf-8-sig"
)

print("✅ messages_clean.csv saved successfully")
```
