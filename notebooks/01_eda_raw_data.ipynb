```python

json_path = r"D:\uni\term9\datamaining\almossyar-data-mining-main\data\raw\data.json"

# لود فایل خام
with open(json_path, encoding="utf-8") as f:
    raw_data = json.load(f)

# تبدیل به DataFrame اولیه
df_raw = pd.DataFrame(raw_data)

# بررسی نوع messages
print(type(df_raw['messages'].iloc[0]))  # دیکشنری یا لیست؟

# استخراج پیام‌ها
all_messages = []

for idx, row in df_raw.iterrows():
    msgs = row['messages']
    
    # اگر messages یه دیکشنری هست
    if isinstance(msgs, dict):
        msgs = [msgs]
    
    for msg in msgs:
        # فقط پیام‌های معمولی
        if msg.get('type') == 'message':
            all_messages.append(msg)

# ساخت DataFrame نهایی
df_messages = pd.json_normalize(all_messages)

# نمایش ستون‌ها و چند نمونه
print(df_messages.columns)
print(df_messages.head())

# ستون متن واقعی
TEXT_COL = 'text'  # اگر ستون متن متفاوت بود، اینو عوض کنید
df_messages['text_length'] = df_messages[TEXT_COL].fillna("").astype(str).apply(len)
df_messages[['text', 'text_length']].head()
```

```python

# ستون متن پیام
TEXT_COL = "text"

# فیلدهای اصلی مورد نیاز پروژه
df_messages_clean = pd.DataFrame({
    "message_id": df_messages["id"],
    "date": df_messages["date"],
    "from_channel": df_messages["from"],
    "text": df_messages[TEXT_COL].fillna("").astype(str),
})

# طول متن برای بررسی اولیه
df_messages_clean["text_length"] = df_messages_clean["text"].apply(len)

# نمایش چند نمونه
df_messages_clean.head()
```
